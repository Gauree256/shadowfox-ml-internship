```python
from transformers import pipeline

sentiment_analyzer = pipeline(
    "sentiment-analysis",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english"
)

print("Sentiment analysis model loaded successfully!")
```

```python
sentences = [
    "This movie was absolutely wonderful!",
    "I hated this movie. It was terrible.",
    "The acting was excellent and the story was amazing.",
    "The movie was boring and disappointing."
]

for sentence in sentences:
    result = sentiment_analyzer(sentence)
    print("Sentence:", sentence)
    print("Result:", result[0]["label"])
    print("Confidence:", round(result[0]["score"] * 100, 2), "%")
    print("-" * 50)
```

```python
sentence = input("Enter a sentence or review: ")

result = sentiment_analyzer(sentence)[0]

print("\nYour sentence:", sentence)
print("Predicted sentiment:", result["label"])
print("Confidence:", round(result["score"] * 100, 2), "%")
```

# Advanced Language Model Project: Sentiment Analysis Using DistilBERT

## 1. Introduction
This project demonstrates how a pretrained transformer-based language model can identify sentiment in English text. Sentiment analysis is the process of classifying text according to the opinion or emotion it expresses.

## 2. Objective
The objective is to use a pretrained DistilBERT model to classify text as POSITIVE or NEGATIVE and examine its predictions and confidence scores.

## 3. Model Used
DistilBERT is a smaller, more efficient model based on BERT. In this project, we use a version fine-tuned for sentiment analysis. The pretrained model allows us to perform sentiment classification without training a language model from scratch.

## 4. Tools and Libraries
- Python
- Google Colab and Jupyter Notebook
- Hugging Face Transformers
- PyTorch

## 5. Methodology
1. Install the required libraries.
2. Load the pretrained sentiment-analysis model.
3. Test the model on example sentences.
4. Accept user-provided text and predict its sentiment.
5. Examine predictions, confidence scores, capabilities, and limitations.

## 6. Initial Observations
The model correctly classified the positive and negative example sentences tested so far. These initial examples demonstrate how the model works, but they are not sufficient to establish its overall accuracy.

## 7. Limitations
The model may struggle with sarcasm, mixed opinions, ambiguous language, or text that differs from its training data. Confidence scores should not be interpreted as guarantees of correctness.

```python

test_sentences = [
    "I love this product. It is fantastic!",
    "This is the worst experience I have ever had.",
    "The service was excellent.",
    "I am very disappointed with this purchase.",
    "The food was delicious.",
    "The food was awful."
]

expected_labels = [
    "POSITIVE",
    "NEGATIVE",
    "POSITIVE",
    "NEGATIVE",
    "POSITIVE",
    "NEGATIVE"
]

predicted_labels = []

for sentence in test_sentences:
    result = sentiment_analyzer(sentence)[0]
    predicted_labels.append(result["label"])

    print("Sentence:", sentence)
    print("Expected:", expected_labels[len(predicted_labels) - 1])
    print("Predicted:", result["label"])
    print("-" * 50)

correct = sum(
    expected == predicted
    for expected, predicted in zip(expected_labels, predicted_labels)
)

accuracy = correct / len(expected_labels) * 100

print("Correct predictions:", correct, "out of", len(expected_labels))
print("Accuracy on these six examples:", round(accuracy, 2), "%")
```

## 8. Evaluation and Results

The model was tested on six example sentences with predefined expected sentiment labels.

- Total test examples: 6
- Correct predictions: 6
- Incorrect predictions: 0
- Accuracy on this small example set: 100%

### Analysis

All six predictions matched their expected sentiment labels. The model correctly classified positive examples such as "The service was excellent" and negative examples such as "The food was awful."

However, this small, manually selected set does not represent all possible English text. Therefore, the observed 100% accuracy should not be considered the model's general performance on real-world data.

A more reliable evaluation would require a larger, representative, labeled test dataset that was not used to select or tune the model.

## 9. Capabilities of the Model

The pretrained DistilBERT sentiment-analysis model has the following capabilities:

1. **Sentiment classification:** Identifies text as POSITIVE or NEGATIVE.
2. **Context understanding:** Uses surrounding words to interpret the meaning of text.
3. **Confidence scores:** Provides a score indicating how strongly the model favors its predicted label.
4. **Interactive predictions:** Allows users to enter their own sentences for analysis.
5. **Efficient inference:** DistilBERT is a smaller model designed to be more efficient than the full BERT model.

## 10. Limitations of the Model

1. **Limited sentiment categories:** The selected model predicts only POSITIVE or NEGATIVE, not a separate NEUTRAL category.
2. **Sarcasm:** It may misunderstand sarcastic sentences.
3. **Ambiguous language:** Sentences with mixed opinions can be difficult to classify.
4. **Language limitations:** This model is intended for English text and may not work reliably for other languages.
5. **Confidence is not certainty:** A high confidence score does not guarantee a correct prediction.
6. **Limited evaluation:** Our six-example evaluation set is too small to establish the model's general accuracy.
